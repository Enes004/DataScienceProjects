```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.columns
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,7))
corrs = df.corr(numeric_only=True)
sns.heatmap(data=corrs , annot = True ,fmt=".2f", cmap="coolwarm",linecolor="black")
plt.tight_layout()
plt.show()
```

```python
# Survived has correlation with adult_male and pclass
# Pclass has correlation with fare and age. 
```

```python
plt.figure(figsize=(7,6))
sns.countplot(x="survived",hue="sex",data=df)
plt.tight_layout()
plt.show()
```

```python
 # KATEGORİK VS KATEGORİK (CROSSTAB X COUNTPLOT+HUE)
plt.figure(figsize=(7,6))
crosstab = pd.crosstab(df["survived"],df["sex"])
sns.heatmap(data=crosstab,annot=True,fmt="d" )
```

```python
crosstab2 = pd.crosstab(df["survived"],df["pclass"])
sns.heatmap(data=crosstab2,fmt="d",annot=True)
```

```python
sns.countplot(data=df,x="pclass",hue="survived")
```

```python
sns.countplot(data=df,x="survived",hue="pclass")
```

```python
# KATEGORİK VS SAYISAL(BOXPLOT/VIOLINPLOT) + 3. BOYUT SURVIVED AGE + SEX
sns.boxplot(data=df,x="survived",y="age",hue="sex")

```

```python
sns.violinplot(data=df,x="survived",y="age",hue="sex",split=True)
```

```python
#sayısal vs sayısal
sns.scatterplot(x="fare",y="age",data=df)
```
